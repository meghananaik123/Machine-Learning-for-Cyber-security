```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder

# 1. Load dataset
df = pd.read_csv(r"C:\Users\Hp\Downloads\wsn_malicious_events_dataset.csv")

# 2. Extract timestamp features
df['Hour'] = pd.to_datetime(df['timestamp']).dt.hour

# 3. Encode source IP and event type
ip_encoder = LabelEncoder()
event_encoder = LabelEncoder()

df['IP_Code'] = ip_encoder.fit_transform(df['source_ip'])
df['Event_Code'] = event_encoder.fit_transform(df['event_type'])

# 4. Select features and target
X = df[['Hour', 'IP_Code', 'Event_Code',
        'packet_count', 'failed_attempts']]
y = df['label']

# 5. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# 6. Train Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 7. Predict results
y_pred = model.predict(X_test)

# 8. Display accuracy and evaluation
print("Accuracy:", accuracy_score(y_test, y_pred) * 100, "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# 9. Display predictions
df['Prediction'] = model.predict(X)

df['Result'] = df['Prediction'].map({
    0: 'Normal Event',
    1: 'Malicious Event'
})

print("\nWSN Event Detection Results:")
print(df[['timestamp', 'source_ip',
          'event_type', 'Result']].head(15))
```
