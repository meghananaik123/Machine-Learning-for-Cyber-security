```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. Generate simulated network traffic data
dates = pd.date_range(start="2024-01-01", periods=120, freq="D")
trend = np.linspace(1000, 1800, 120)
seasonal = 200 * np.sin(np.arange(120) * 2 * np.pi / 7)
noise = np.random.normal(0, 50, 120)
traffic = trend + seasonal + noise

# 2. Create the DataFrame and set the Date index
data = pd.DataFrame({
    "Date": dates,
    "Network_Traffic": traffic
})
data.set_index("Date", inplace=True)

# 3. Apply seasonal decomposition (Additive Model with a weekly period)
result = seasonal_decompose(data["Network_Traffic"], model="additive", period=7)

# 4. Plot the decomposed components
fig = result.plot()
fig.set_size_inches(10, 8)  # Makes the graph clear and readable

# Add title with proper layout adjustment to prevent overlapping
plt.suptitle("Time Series Decomposition of Network Traffic", fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```
